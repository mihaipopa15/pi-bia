```python
#ex1
from sklearn.linear_model import SGDRegressor
import numpy as np
from sklearn.datasets import make_regression
from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_squared_error
```

```python
X, y = make_regression(n_samples=1000, n_features=1, noise=0.5, bias=3,  random_state=42)
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.15, random_state=0)
```

```python
sgd_reg = SGDRegressor(max_iter=150, alpha=0.01, penalty='l2', eta0=0.005, random_state=42)
sgd_reg.fit(X, y.ravel())
```

```python
sgd_reg.intercept_, sgd_reg.coef_
```

```python
y_train_predict = sgd_reg.predict(X_test)
```

```python
mean_squared_error(y_test, y_train_predict)
```

```python
from sklearn.datasets import make_classification
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier
from sklearn.ensemble import VotingClassifier
from sklearn.linear_model import LogisticRegression
from sklearn.svm import SVC
```

```python
X,y = make_classification(n_samples=1000, n_features=4, n_informative=4, n_redundant=0,
                          n_classes=3, n_clusters_per_class=1, random_state=42)
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.25, random_state=42)
```

```python
log_clf = LogisticRegression(solver="lbfgs", random_state=42)
rnd_clf = RandomForestClassifier(n_estimators=200, random_state=42)
svm_clf = SVC(gamma=1.5, probability=True, random_state=42)
```

```python
voting_clf = VotingClassifier(
    estimators=[('lr', log_clf), ('rf', rnd_clf), ('svc', svm_clf)],
    voting='soft')
voting_clf.fit(X_train, y_train)
```

```python
voting_clf.score(X_test, y_test)
```

```python
sample3 = voting_clf.predict_proba(X_test)
sample3[8]
```

```python
from sklearn.datasets import make_circles
from sklearn.model_selection import train_test_split
from sklearn.decomposition import PCA
```

```python
X,y = make_circles(n_samples=1000, factor=0.1, noise=0.01, random_state=42)
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.25, random_state=42)
```

```python
from sklearn.decomposition import PCA
pca = PCA(n_components=0.95)
pca.fit_transform(X_train)
```

```python
num = pca.n_components_
num
```

```python
comp = pca.components_
comp
```

```python
pca.explained_variance_ratio_
```

```python
sample_8_test = X_test[7].reshape(1, -1)
sample_8_pca = pca.transform(sample_8_test)
inverse_transformation = pca.inverse_transform(sample_8_pca)[0]
inverse_transformation
```

```python
X, y = make_circles(n_samples=1000, factor=0.5, noise=0.1, random_state=42)
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.25, random_state=42)
```

```python
from sklearn.cluster import KMeans
kmeans = KMeans(n_clusters=3, random_state=42, n_init="auto")
y_pred = kmeans.fit_predict(X_train)
```

```python
label_11th_sample = kmeans.predict(X_test[10].reshape(1, -1))[0]
label_11th_sample
```

```python
first_cluster_center = kmeans.cluster_centers_[0]
first_cluster_center
```

```python
inertia_train = kmeans.inertia_
inertia_train
```

```python
negative_inertia_test = kmeans.score(X_test)
negative_inertia_test
```

```python
from sklearn.metrics import normalized_mutual_info_score
y_pred_test = kmeans.predict(X_test)
nmi_score = normalized_mutual_info_score(y_test, y_pred_test)
nmi_score
```

```python
#ex2
from sklearn.svm import SVR
from sklearn.metrics import mean_squared_error
```

```python
X, y = make_regression(n_samples=1000, n_features=2, noise=0.5, bias=1, random_state=42)
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.25, random_state=42)
```

```python
svm_regressor = SVR(kernel='linear', C=0.1, epsilon=0.5, max_iter=500)
svm_regressor.fit(X_train, y_train)
```

```python
w = svm_regressor.coef_
w
```

```python
b = svm_regressor.intercept_
b
```

```python
first_support_vector = svm_regressor.support_vectors_[0]
first_support_vector
```

```python
y_pred = svm_regressor.predict(X_test)
mse = mean_squared_error(y_test, y_pred)
mse
```
